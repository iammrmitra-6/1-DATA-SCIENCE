```python
import matplotlib.pyplot as plt
x=[1,2,3,5,7]
y=[1,4,9,25,49]

##create a line plot
plt.plot(x,y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('basic line plot')
plt.show()
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

### create acutomized line plot

plt.plot(x,y,color='red',linestyle='-.',marker="o",linewidth=3,markersize=9)
plt.grid(True)
```

```python
## multiple plots
## sample data
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color="green")
plt.title('plot 1')

plt.subplot(2,2,2)
plt.plot(y1,x,color="red")
plt.title('plot 2')

plt.subplot(2,2,3)
plt.plot(x,y2,color="blue")
plt.title('plot 3')

plt.subplot(2,2,4)
plt.plot(y1,y2,color="blue")
plt.title('plot 4')
```

```python
#bar plor
categories=['A',"B","C",'D','E']
values=[5,7,3,8,6]

plt.bar(categories,values,color="purple",width=0.34)
plt.xlabel('categories')
plt.ylabel("values")
plt.title('bar plot')
plt.show()
```
