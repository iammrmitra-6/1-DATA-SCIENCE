```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

   
    def bark(self):
        print(f"{self.name} says woof")

dog1 = Dog("Rocky", 6)
dog1.bark()
dog2= Dog("lucy",3)
dog2.bark()

```

```python
#inheritance

#parent class
class car:
    def   __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self):
        print(f"the person will drive {self.enginetype} car")
```

```python
car1=car(4,5,"petrol")
car1.drive()
```

```python
class tesla(car):
    def __init__(self,windows,doors,enginetype,is_selfdrive):
        super().__init__(windows,doors,enginetype)
        self.is_selfdrive=is_selfdrive
    def selfdriving(self):
        print(f"tesla supports self driving {self.is_selfdrive}")

tesla1= tesla(4,5,"electric",True)
tesla1.selfdriving()
tesla1.drive()
```

```python
# Parent Class
class Animal:
    def speak(self):
        return "sound of the animal"

# Standalone helper function to demonstrate polymorphism
def animal_speak(animal_obj):
    print(animal_obj.speak())

# Child Class 1
class Dog(Animal):
    def speak(self):
        return "woof!"

# Child Class 2
class Cat(Animal):
    def speak(self):  # Fixed 'Self' to 'self'
        return "meowwwww"

# --- Execution ---

# 1. Create instances
dog1 = Dog()
cat1 = Cat()  # Created cat1

# 2. Testing Method Overriding
print(dog1.speak())       # Output: woof!
print(cat1.speak())       # Output: meowwwww

# 3. Testing Polymorphism
animal_speak(dog1)        # Output: woof!
animal_speak(cat1)        # Output: meowwwww
```

```python
##closures


def main_welcome():
    msg="hello world"
    def inner_welcome():
        print("hello python world")
        print(msg) ##in closure function we can acesss anything we declare outside the inner function
    return inner_welcome()

main_welcome()
```

```python
# 1. Define the decorator
def main_welcome(func):
    msg = "hello world"
    
    def inner_welcome():
        print("hello python world")
        func()  # This calls the original function we passed in
        print(msg) 
        
    return inner_welcome  # <--- Note: No parenthesis here! We return the function itself.

# 2. Define the base function
def greet():
    print("hello shannidhya")

# 3. Apply the decorator manually
decorated_greet = main_welcome(greet)  # <--- Note: We pass 'greet', not 'greet()'

# 4. Run the decorated function
decorated_greet()
```

```python
@main_welcome
def coure_introduction():
    print("this is an advanced python course")
coure_introduction
```
